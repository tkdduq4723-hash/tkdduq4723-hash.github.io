```python
import requests
import pandas as pd
import re
import math

# API키 입력!
NAVER_CLIENT_ID = "----"
NAVER_CLIENT_SECRET = "----"
KAKAO_API_KEY = "----"

PNU_LAT = 35.2323
PNU_LNG = 129.0847

def get_naver_restaurants(query="부산대 맛집", display=5):
    url = "https://openapi.naver.com/v1/search/local.json"
    headers = {
        "X-Naver-Client-Id": NAVER_CLIENT_ID,
        "X-Naver-Client-Secret": NAVER_CLIENT_SECRET
    }
    params = {"query": query, "display": display, "sort": "comment"}
    res = requests.get(url, headers=headers, params=params)
    return res.json().get("items", [])

def get_kakao_restaurants(lat=PNU_LAT, lng=PNU_LNG, radius=500):
    url = "https://dapi.kakao.com/v2/local/search/category.json"
    headers = {"Authorization": f"KakaoAK {KAKAO_API_KEY}"}
    params = {
        "category_group_code": "FD6",
        "x": lng,
        "y": lat,
        "radius": radius,
        "sort": "distance",
        "size": 15
    }
    res = requests.get(url, headers=headers, params=params)
    return res.json().get("documents", [])

def calc_walk_time(distance_m):
    if not distance_m:
        return "-"

    real_distance = int(distance_m) * 1.3
    minutes = real_distance / 67

    if minutes < 1:
        return "1분 이내"

    return f"약 {math.ceil(minutes)}분"

naver_data = get_naver_restaurants()
kakao_data = get_kakao_restaurants()

naver_df = pd.DataFrame([{
    "이름": re.sub('<.*?>', '', r["title"]),
    "주소": r["roadAddress"],
    "카테고리": r["category"],
    "거리(m)": "-",
    "도보시간": "-",
    "출처": "네이버"
} for r in naver_data])

kakao_df = pd.DataFrame([{
    "이름": r["place_name"],
    "주소": r["road_address_name"],
    "카테고리": r["category_name"],
    "거리(m)": r["distance"],
    "도보시간": calc_walk_time(r["distance"]),
    "출처": "카카오"
} for r in kakao_data])

df = pd.concat([naver_df, kakao_df], ignore_index=True)
df
```

```python
df.to_excel("부산대_식당목록.xlsx", index=False)

# 코랩에서 자동 다운로드
from google.colab import files
files.download("부산대_식당목록.xlsx")
```
